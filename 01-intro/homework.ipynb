```python
import pandas as pd
import numpy as np

data = 'https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv'
!wget $data

df = pd.read_csv(data)
df.head()
```

```python
# Q1
print(pd.__version__)

# Q2
print(len(df))

# Q3
print(df.fuel_type.unique().size)

# Q4
print(df.isna().sum())

# Q5
print(df[df['origin'] == 'Asia'].fuel_efficiency_mpg.max())

# Q6
median_val_hp = df.horsepower.median()
most_frequent_val = df.horsepower.value_counts().index[0]
df.horsepower = df.horsepower.fillna(most_frequent_val)
new_med_val_hp = df.horsepower.median()
print(new_med_val_hp - median_val_hp)

# Q7
df_asian = df[df['origin'] == 'Asia']
df1 = df_asian[['vehicle_weight','model_year']]
df2 = df1.iloc[:7]
X = df2.to_numpy()
X_T = X.T

XTX = X_T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = XTX_inv.dot(X_T).dot(y)
print(w.sum().round(3))
```
